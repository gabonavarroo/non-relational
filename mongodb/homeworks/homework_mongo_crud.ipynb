## Sección 7 - Estructuras complejas y Agregación

La siguiente celda es de solo lectura e insertará una lista de clientes que será requerida para las preguntas posteriores

````python
import random
import faker
import faker_commerce
from mimesis import Generic
from datetime import datetime, timedelta, timezone
from typing import cast
from pprint import pprint

fake = faker.Faker("es_MX")
fake.add_provider(faker_commerce.Provider)
mim_generic = Generic(locale="es_MX")

#######################

productos = []
for _ in range(random.randint(50, 100)):
    producto = {
        "producto": cast(faker_commerce.Provider, fake.unique).ecommerce_name(),
        "categoria": cast(faker_commerce.Provider, fake).ecommerce_category(),
        "precio": cast(faker_commerce.Provider, fake).ecommerce_price(as_int=False)
        / 5000.0,
    }
    productos.append(producto)

database_name = "laboratorio"
collection_name = "productos"
mongo_client[database_name].drop_collection(collection_name)
mongo_client[database_name][collection_name].insert_many(productos)
print(f'{len(productos)} documentos insertados en colección "{collection_name}"')

#######################

clientes = []
for _ in range(random.randint(500, 1000)):
    client = {
        "nombre": cast(faker.Faker, fake.unique).name(),
        "edad": random.randint(18, 65),
        "contacto": {
            "telefono": cast(faker.Faker, fake.unique).phone_number(),
            "email": cast(faker.Faker, fake.unique).email(),
        },
        "compras": [],
    }
    for _ in range(random.randint(1, 5)):
        producto = random.choice(productos)
        cantidad = random.randint(1, 3)
        compra = {
            "producto_id": producto["_id"],
            "fecha": fake.past_datetime(
                start_date=datetime.now(tz=timezone.utc) - timedelta(days=365)
            ),
            "cantidad": cantidad,
            "monto": producto["precio"] * cantidad,
        }
        client["compras"].append(compra)
    clientes.append(client)

database_name = "laboratorio"
collection_name = "clientes"
mongo_client[database_name].drop_collection(collection_name)
mongo_client[database_name][collection_name].insert_many(clientes)
print(f'{len(clientes)} documentos insertados en colección "{collection_name}"')
````

Clasifica los clientes en buenos o malos.

- En una variable con nombre "buenos_clientes" almacena el resultado de una consulta que devuelva todos los clientes con al menos una compra de monto mayor o igual a 500

- En una variable con nombre "malos_clientes" almacena el resultado de una consulta que devuelva todos los clientes cuyas compras sean todas menores a 500

***En ambos casos devuelve el documento completo (es decir, sin omitir o agregar atributos, sin proyecciones, sin agrupaciones)***

````python
# Clientes con AL MENOS UNA compra con monto >= 500
buenos_clientes = db["clientes"].find(
    {"compras": {"$elemMatch": {"monto": {"$gte": 500}}}}
)

# Clientes donde NINGUNA compra tiene monto >= 500
malos_clientes = db["clientes"].find(
    {"compras": {"$not": {"$elemMatch": {"monto": {"$gte": 500}}}}}
)
````

Crea una consulta que devuelve todos los clientes enriqueciendo la información de sus compras con los datos de la collecion producto.

Es decir, cada resultado de la consulta debe seguir el esquema

```json
{
    "_id": ...,
    "compras": [
        {
            "_id": ...,
            "cantidad": ...,
            "categoria": ...,
            "fecha": ...,
            "monto": ...,
            "precio": ...,
            "producto": ...,
            "producto_id": ...
        },
        ...
    ],
    "contacto": {
        "email": ...,
        "telefono": ...
    },
    "edad": ...,
    "nombre": ...
}
```

***Almacena los resultados de la consulta en una variable con nombre "compras_enriquecidas"***

````python
compras_enriquecidas = db["clientes"].aggregate([

    {"$unwind": "$compras"},

    {"$lookup": {
        "from": "productos",
        "localField": "compras.producto_id",
        "foreignField": "_id",
        "as": "producto_info"
    }},

    {"$unwind": "$producto_info"},

    {"$addFields": {
        "compras": {
            "$mergeObjects": [
                "$compras",
                {
                    "_id": "$producto_info._id",
                    "producto": "$producto_info.producto",
                    "categoria": "$producto_info.categoria",
                    "precio": "$producto_info.precio"
                }
            ]
        }
    }},

    {"$group": {
        "_id": "$_id",
        "nombre": {"$first": "$nombre"},
        "edad": {"$first": "$edad"},
        "contacto": {"$first": "$contacto"},
        "compras": {"$push": "$compras"}
    }}
])
````
